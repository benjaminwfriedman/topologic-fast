# Floorplan to Graph Visualization

This notebook demonstrates:
1. Creating a floorplan of rooms (cells)
2. Combining them into a CellComplex
3. Creating a dual graph (rooms as vertices, shared walls as edges)
4. Visualizing both the geometry and the graph

````python
import topologic_fast as tf
import plotly.graph_objects as go
from plotly.subplots import make_subplots
````

## 1. Create a Simple Floorplan

We'll create a simple floorplan with 6 rooms:
```
+-------+-------+-------+
|       |       |       |
| Room4 | Room5 | Room6 |
|       |       |       |
+-------+-------+-------+
|       |       |       |
| Room1 | Room2 | Room3 |
|       |       |       |
+-------+-------+-------+
```

````python
# Room dimensions
room_width = 4.0
room_length = 3.0
room_height = 2.5

# Create rooms (cells) - bottom row
room1 = tf.Cell.Box(0, 0, 0, room_width, room_length, room_height)
room2 = tf.Cell.Box(room_width, 0, 0, room_width, room_length, room_height)
room3 = tf.Cell.Box(2 * room_width, 0, 0, room_width, room_length, room_height)

# Create rooms - top row
room4 = tf.Cell.Box(0, room_length, 0, room_width, room_length, room_height)
room5 = tf.Cell.Box(room_width, room_length, 0, room_width, room_length, room_height)
room6 = tf.Cell.Box(2 * room_width, room_length, 0, room_width, room_length, room_height)

rooms = [room1, room2, room3, room4, room5, room6]
room_names = ['Room 1', 'Room 2', 'Room 3', 'Room 4', 'Room 5', 'Room 6']

print(f"Created {len(rooms)} rooms")
for i, room in enumerate(rooms):
    print(f"  {room_names[i]}: Volume = {room.Volume():.1f} m³")
````

## 2. Create CellComplex from Rooms

````python
# Combine all rooms into a CellComplex
floorplan = tf.CellComplex.ByCells(rooms)

print(f"CellComplex created:")
print(f"  Cells: {floorplan.NumCells()}")
print(f"  Total Volume: {floorplan.Volume():.1f} m³")
print(f"  Total Area: {floorplan.Area():.1f} m²")
````

## 3. Create Dual Graph

The dual graph represents:
- Each room as a vertex (at the room's centroid)
- Each shared wall as an edge connecting adjacent rooms

````python
# Create the dual graph
graph = tf.Graph.ByTopology(floorplan)

print(f"Dual Graph created:")
print(f"  Vertices (rooms): {graph.Order()}")
print(f"  Edges (shared walls): {graph.Size()}")

# Get graph properties
print(f"\nGraph properties:")
print(f"  Density: {graph.Density():.3f}")
print(f"  Diameter: {graph.Diameter()}")
print(f"  Is Bipartite: {graph.IsBipartite()}")
````

## 4. Visualize the CellComplex

````python
def visualize_cellcomplex(cellcomplex, title="CellComplex"):
    """Create a 3D visualization of a CellComplex"""
    fig = go.Figure()
    
    # Colors for different rooms
    colors = ['lightblue', 'lightgreen', 'lightyellow', 'lightpink', 'lightcoral', 'lightsalmon']
    
    cells = cellcomplex.Cells()
    
    for i, cell in enumerate(cells):
        color = colors[i % len(colors)]
        faces = cell.Faces()
        
        for face in faces:
            # Get vertices for the face
            vertices = face.Vertices()
            coords = [v.Coordinates() for v in vertices]
            
            if len(coords) >= 3:
                # Create mesh for each face
                x = [c[0] for c in coords]
                y = [c[1] for c in coords]
                z = [c[2] for c in coords]
                
                # Add face as mesh
                fig.add_trace(go.Mesh3d(
                    x=x, y=y, z=z,
                    color=color,
                    opacity=0.5,
                    alphahull=0,
                    name=f'Room {i+1}',
                    showlegend=(face == faces[0])  # Only show legend once per room
                ))
                
                # Add edges
                for j in range(len(coords)):
                    p1 = coords[j]
                    p2 = coords[(j + 1) % len(coords)]
                    fig.add_trace(go.Scatter3d(
                        x=[p1[0], p2[0]], y=[p1[1], p2[1]], z=[p1[2], p2[2]],
                        mode='lines',
                        line=dict(color='black', width=2),
                        showlegend=False
                    ))
    
    fig.update_layout(
        title=title,
        scene=dict(
            aspectmode='data',
            xaxis_title='X (m)',
            yaxis_title='Y (m)',
            zaxis_title='Z (m)'
        ),
        width=800,
        height=600
    )
    
    return fig

fig_complex = visualize_cellcomplex(floorplan, "Floorplan - 3D View")
fig_complex.show()
````

## 5. Visualize the Dual Graph

````python
def visualize_graph(graph, room_names=None, title="Dual Graph"):
    """Create a 3D visualization of the graph"""
    fig = go.Figure()
    
    # Get vertices and edges
    vertices = graph.Vertices()
    edges = graph.Edges()
    
    # Extract vertex coordinates
    vertex_coords = [v.Coordinates() for v in vertices]
    
    # Add edges as lines
    for edge in edges:
        edge_verts = edge.Vertices()
        if len(edge_verts) == 2:
            p1 = edge_verts[0].Coordinates()
            p2 = edge_verts[1].Coordinates()
            fig.add_trace(go.Scatter3d(
                x=[p1[0], p2[0]],
                y=[p1[1], p2[1]],
                z=[p1[2], p2[2]],
                mode='lines',
                line=dict(color='blue', width=5),
                showlegend=False,
                hoverinfo='skip'
            ))
    
    # Add vertices as markers
    x = [c[0] for c in vertex_coords]
    y = [c[1] for c in vertex_coords]
    z = [c[2] for c in vertex_coords]
    
    labels = room_names if room_names else [f'Room {i+1}' for i in range(len(vertices))]
    
    fig.add_trace(go.Scatter3d(
        x=x, y=y, z=z,
        mode='markers+text',
        marker=dict(size=15, color='red', symbol='circle'),
        text=labels,
        textposition='top center',
        textfont=dict(size=12, color='black'),
        name='Rooms',
        hovertemplate='%{text}<br>x: %{x:.1f}<br>y: %{y:.1f}<br>z: %{z:.1f}<extra></extra>'
    ))
    
    fig.update_layout(
        title=title,
        scene=dict(
            aspectmode='data',
            xaxis_title='X (m)',
            yaxis_title='Y (m)',
            zaxis_title='Z (m)'
        ),
        width=800,
        height=600
    )
    
    return fig

fig_graph = visualize_graph(graph, room_names, "Room Connectivity Graph")
fig_graph.show()
````

## 6. Combined Visualization

````python
def visualize_combined(cellcomplex, graph, room_names=None, title="Floorplan with Connectivity Graph"):
    """Overlay the graph on the CellComplex visualization"""
    fig = go.Figure()
    
    colors = ['rgba(173,216,230,0.3)', 'rgba(144,238,144,0.3)', 'rgba(255,255,224,0.3)', 
              'rgba(255,182,193,0.3)', 'rgba(240,128,128,0.3)', 'rgba(255,160,122,0.3)']
    
    cells = cellcomplex.Cells()
    
    # Draw CellComplex faces (transparent)
    for i, cell in enumerate(cells):
        color = colors[i % len(colors)]
        faces = cell.Faces()
        
        for face in faces:
            vertices = face.Vertices()
            coords = [v.Coordinates() for v in vertices]
            
            if len(coords) >= 3:
                x = [c[0] for c in coords]
                y = [c[1] for c in coords]
                z = [c[2] for c in coords]
                
                fig.add_trace(go.Mesh3d(
                    x=x, y=y, z=z,
                    color=color,
                    opacity=0.3,
                    alphahull=0,
                    showlegend=False
                ))
                
                # Add edges
                for j in range(len(coords)):
                    p1 = coords[j]
                    p2 = coords[(j + 1) % len(coords)]
                    fig.add_trace(go.Scatter3d(
                        x=[p1[0], p2[0]], y=[p1[1], p2[1]], z=[p1[2], p2[2]],
                        mode='lines',
                        line=dict(color='gray', width=1),
                        showlegend=False,
                        hoverinfo='skip'
                    ))
    
    # Draw graph edges (thick blue lines)
    graph_vertices = graph.Vertices()
    graph_edges = graph.Edges()
    
    for edge in graph_edges:
        edge_verts = edge.Vertices()
        if len(edge_verts) == 2:
            p1 = edge_verts[0].Coordinates()
            p2 = edge_verts[1].Coordinates()
            fig.add_trace(go.Scatter3d(
                x=[p1[0], p2[0]],
                y=[p1[1], p2[1]],
                z=[p1[2], p2[2]],
                mode='lines',
                line=dict(color='blue', width=8),
                showlegend=False,
                hoverinfo='skip'
            ))
    
    # Draw graph vertices (large red markers)
    vertex_coords = [v.Coordinates() for v in graph_vertices]
    x = [c[0] for c in vertex_coords]
    y = [c[1] for c in vertex_coords]
    z = [c[2] for c in vertex_coords]
    
    labels = room_names if room_names else [f'Room {i+1}' for i in range(len(graph_vertices))]
    
    fig.add_trace(go.Scatter3d(
        x=x, y=y, z=z,
        mode='markers+text',
        marker=dict(size=12, color='red', symbol='circle'),
        text=labels,
        textposition='top center',
        textfont=dict(size=10, color='darkblue'),
        name='Room Centers',
        hovertemplate='%{text}<extra></extra>'
    ))
    
    fig.update_layout(
        title=title,
        scene=dict(
            aspectmode='data',
            xaxis_title='X (m)',
            yaxis_title='Y (m)',
            zaxis_title='Z (m)',
            camera=dict(eye=dict(x=1.5, y=1.5, z=1.0))
        ),
        width=900,
        height=700
    )
    
    return fig

fig_combined = visualize_combined(floorplan, graph, room_names, "Floorplan with Room Connectivity")
fig_combined.show()
````

## 7. Graph Analysis - Pathfinding

Find the shortest path between rooms using the dual graph.

````python
# Get graph vertices (room centroids)
graph_vertices = graph.Vertices()

# Find shortest path from Room 1 to Room 6 (opposite corners)
start_room = graph_vertices[0]  # Room 1
end_room = graph_vertices[5]    # Room 6

distance = graph.Distance(start_room, end_room)
print(f"Distance from Room 1 to Room 6: {distance} rooms")

# Find the path
path = graph.Path(start_room, end_room)
if path:
    path_vertices = path.Vertices()
    print(f"\nShortest path ({len(path_vertices)} rooms):")
    for i, v in enumerate(path_vertices):
        # Find which room this vertex corresponds to
        coords = v.Coordinates()
        for j, gv in enumerate(graph_vertices):
            gc = gv.Coordinates()
            if abs(coords[0] - gc[0]) < 0.1 and abs(coords[1] - gc[1]) < 0.1:
                print(f"  Step {i+1}: {room_names[j]}")
                break
````

## 8. Adjacency Analysis

````python
# Get adjacency matrix
adj_matrix = graph.AdjacencyMatrix()

print("Room Adjacency Matrix:")
print("        ", "  ".join([f"R{i+1}" for i in range(len(room_names))]))
for i, row in enumerate(adj_matrix):
    print(f"Room {i+1}: ", "   ".join([str(x) for x in row]))

print("\nRoom Connections:")
for i, vertex in enumerate(graph_vertices):
    adjacent = graph.AdjacentVertices(vertex)
    adjacent_names = []
    for adj_v in adjacent:
        adj_coords = adj_v.Coordinates()
        for j, gv in enumerate(graph_vertices):
            gc = gv.Coordinates()
            if abs(adj_coords[0] - gc[0]) < 0.1 and abs(adj_coords[1] - gc[1]) < 0.1:
                adjacent_names.append(room_names[j])
                break
    print(f"  {room_names[i]} connects to: {', '.join(adjacent_names)}")
````

## 9. 2D Floor Plan View

````python
def visualize_2d_floorplan(cellcomplex, graph, room_names=None):
    """Create a 2D top-down view of the floorplan with graph overlay"""
    fig = go.Figure()
    
    colors = ['lightblue', 'lightgreen', 'lightyellow', 'lightpink', 'lightcoral', 'lightsalmon']
    
    cells = cellcomplex.Cells()
    
    # Draw room outlines (only bottom faces)
    for i, cell in enumerate(cells):
        faces = cell.Faces()
        # Find the bottom face (z = 0)
        for face in faces:
            vertices = face.Vertices()
            coords = [v.Coordinates() for v in vertices]
            # Check if this is a horizontal face at z=0
            z_coords = [c[2] for c in coords]
            if all(abs(z) < 0.01 for z in z_coords):
                x = [c[0] for c in coords] + [coords[0][0]]
                y = [c[1] for c in coords] + [coords[0][1]]
                
                fig.add_trace(go.Scatter(
                    x=x, y=y,
                    fill='toself',
                    fillcolor=colors[i % len(colors)],
                    line=dict(color='black', width=2),
                    name=room_names[i] if room_names else f'Room {i+1}',
                    hoverinfo='name'
                ))
                break
    
    # Draw graph edges
    graph_edges = graph.Edges()
    for edge in graph_edges:
        edge_verts = edge.Vertices()
        if len(edge_verts) == 2:
            p1 = edge_verts[0].Coordinates()
            p2 = edge_verts[1].Coordinates()
            fig.add_trace(go.Scatter(
                x=[p1[0], p2[0]],
                y=[p1[1], p2[1]],
                mode='lines',
                line=dict(color='blue', width=4),
                showlegend=False,
                hoverinfo='skip'
            ))
    
    # Draw graph vertices
    graph_vertices = graph.Vertices()
    vertex_coords = [v.Coordinates() for v in graph_vertices]
    x = [c[0] for c in vertex_coords]
    y = [c[1] for c in vertex_coords]
    labels = room_names if room_names else [f'R{i+1}' for i in range(len(graph_vertices))]
    
    fig.add_trace(go.Scatter(
        x=x, y=y,
        mode='markers+text',
        marker=dict(size=20, color='red', line=dict(color='darkred', width=2)),
        text=labels,
        textposition='middle center',
        textfont=dict(size=10, color='white'),
        name='Room Centers',
        hovertemplate='%{text}<extra></extra>'
    ))
    
    fig.update_layout(
        title='2D Floor Plan with Connectivity Graph',
        xaxis=dict(title='X (m)', scaleanchor='y', scaleratio=1),
        yaxis=dict(title='Y (m)'),
        width=800,
        height=500,
        showlegend=True
    )
    
    return fig

fig_2d = visualize_2d_floorplan(floorplan, graph, room_names)
fig_2d.show()
````

## Summary

This notebook demonstrated:

1. **Creating rooms as Cells** - Each room is a 3D box (Cell)
2. **Combining into CellComplex** - All rooms merged into a single topological structure
3. **Creating dual Graph** - `Graph.ByTopology()` creates a graph where:
   - Each Cell (room) becomes a vertex at its centroid
   - Cells sharing a Face (wall) are connected by an edge
4. **Visualization** - 3D and 2D views of both geometry and connectivity
5. **Graph Analysis** - Pathfinding and adjacency analysis

This pattern is useful for:
- Building information modeling (BIM)
- Space syntax analysis
- Navigation and wayfinding
- HVAC and energy modeling
